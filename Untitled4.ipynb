```python
"""
🚀 COMPLETE COMFYUI LAUNCHER
Installs dependencies and starts ComfyUI - all in one
"""

import subprocess
import sys
import time
import os

print("=" * 70)
print("🎨 COMPLETE COMFYUI LAUNCHER")
print("=" * 70)
print()

# ============================================
# PHASE 1: INSTALL DEPENDENCIES
# ============================================
print("📦 PHASE 1: Installing Dependencies")
print("-" * 70)
print()

print("[1/5] 🔧 Fixing NumPy...")
subprocess.run([sys.executable, '-m', 'pip', 'uninstall', '-y', 'numpy'], 
               stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'numpy<2'])
print("✅ NumPy fixed")

print("[2/5] 📥 Installing missing packages...")
missing = ['einops', 'kornia', 'spandrel', 'soundfile', 'pycloudflared']
for pkg in missing:
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', pkg],
                   stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
print("✅ Missing packages installed")

print("[3/5] 📄 Installing from requirements.txt...")
req_file = '/teamspace/studios/this_studio/ComfyUI/requirements.txt'
if os.path.exists(req_file):
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', req_file],
                   stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    print("✅ Requirements installed")
else:
    print("ℹ️  No requirements.txt (that's okay)")

print("[4/5] 🧹 Cleaning up old processes...")
subprocess.run(['pkill', '-f', 'main.py'], 
               stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
time.sleep(2)
print("✅ Ready")

print("[5/5] 📁 Navigating to ComfyUI...")
os.chdir('/teamspace/studios/this_studio/ComfyUI')
print(f"✅ {os.getcwd()}")

print()
print("=" * 70)

# ============================================
# PHASE 2: START COMFYUI
# ============================================
print("🚀 PHASE 2: Starting ComfyUI")
print("-" * 70)
print()

# Start ComfyUI
comfyui_process = subprocess.Popen(
    [sys.executable, 'main.py', '--listen', '0.0.0.0', '--port', '8188'],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1
)

# Monitor startup
print("📊 ComfyUI startup output:")
print()

startup_success = False
line_count = 0

try:
    for line in comfyui_process.stdout:
        print(line, end='')
        line_count += 1
        
        # Check for success indicators
        if any(phrase in line for phrase in [
            'To see the GUI go to:',
            'Running on',
            'Workflow API server listening'
        ]):
            startup_success = True
            print()
            print("-" * 70)
            print("✅ ComfyUI started successfully!")
            print("-" * 70)
            print()
            break
        
        # Check for errors
        if 'ModuleNotFoundError' in line or 'ImportError' in line:
            print()
            print("-" * 70)
            print("❌ Missing dependency detected!")
            print("   Installing and restarting...")
            print("-" * 70)
            comfyui_process.terminate()
            
            # Extract package name from error
            if 'No module named' in line:
                pkg = line.split("'")[1]
                print(f"   Installing {pkg}...")
                subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', pkg])
                print(f"   ✅ {pkg} installed")
                print()
                print("Please run this cell again to restart ComfyUI")
            
            raise SystemExit(0)
        
        # Limit initial output
        if line_count > 40 and not startup_success:
            print()
            print("(Still initializing, output continues in background...)")
            print()
            break
    
    # If we didn't see success message, wait a bit
    if not startup_success:
        print("⏳ Waiting for ComfyUI to fully start...")
        time.sleep(15)
        startup_success = True
    
    print()
    print("=" * 70)
    
    # ============================================
    # PHASE 3: CREATE PUBLIC TUNNEL
    # ============================================
    print("🌐 PHASE 3: Creating Public Access")
    print("-" * 70)
    print()
    
    from pycloudflared import try_cloudflare
    
    print("Creating Cloudflare tunnel...")
    tunnel_url = try_cloudflare(port=8188)
    url = str(tunnel_url)
    
    print()
    print("=" * 70)
    print("=" * 70)
    print()
    print("    🎉 SUCCESS! COMFYUI IS LIVE! 🎉")
    print()
    print("=" * 70)
    print("=" * 70)
    print()
    print("🌍 YOUR PUBLIC URL:")
    print()
    print(f"    {url}")
    print()
    print("=" * 70)
    print()
    print("📋 INSTRUCTIONS:")
    print("   1. Copy the URL above")
    print("   2. Paste it in your browser")
    print("   3. ComfyUI will load!")
    print()
    print("💡 QUICK START:")
    print("   • The interface will show some nodes")
    print("   • Click 'Queue Prompt' to generate an image")
    print("   • Results appear in the right panel")
    print("   • Check output/ folder for saved files")
    print()
    print("🔧 FEATURES:")
    print("   • ComfyUI Manager (click Manager button)")
    print("   • AnimateDiff (video generation)")
    print("   • Multiple custom nodes installed")
    print()
    print("🛑 TO STOP:")
    print("   • Click Stop button in Jupyter")
    print("   • Or press Ctrl+C")
    print()
    print("=" * 70)
    print()
    print("📊 Live ComfyUI logs (this cell keeps running):")
    print("-" * 70)
    print()
    
    # Stream logs
    for line in comfyui_process.stdout:
        print(line, end='')
        
except KeyboardInterrupt:
    print()
    print()
    print("=" * 70)
    print("🛑 Shutting down ComfyUI...")
    comfyui_process.terminate()
    try:
        comfyui_process.wait(timeout=5)
    except:
        comfyui_process.kill()
    print("✅ ComfyUI stopped")
    print("=" * 70)
    
except Exception as e:
    print()
    print(f"⚠️ Error: {e}")
    print()
    print("ComfyUI may still be running at:")
    print("   http://localhost:8188")
    print()
    comfyui_process.terminate()
```
