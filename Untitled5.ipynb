```python
"""
🔧 COMPLETE VERSION FIX
Fixes NumPy, torch, and transformers version conflicts
"""

import subprocess
import sys

print("=" * 70)
print("🔧 FIXING ALL VERSION CONFLICTS")
print("=" * 70)
print()

print("This will take 3-5 minutes...")
print()

# Step 1: Uninstall problematic packages
print("[1/4] 🗑️  Removing conflicting packages...")
packages_to_remove = ['numpy', 'torch', 'torchvision', 'torchaudio', 'transformers']

for pkg in packages_to_remove:
    print(f"   Removing {pkg}...", end=' ')
    subprocess.run(
        [sys.executable, '-m', 'pip', 'uninstall', '-y', pkg],
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL
    )
    print("✅")

print()

# Step 2: Install compatible versions
print("[2/4] 📥 Installing compatible versions...")
print("   This is the longest step (2-3 minutes)...")
print()

compatible_versions = [
    'numpy<2',  # NumPy 1.x
    'torch==2.1.0',  # Stable torch version
    'torchvision==0.16.0',  # Compatible with torch 2.1
    'torchaudio==2.1.0',  # Compatible with torch 2.1
    'transformers==4.35.0',  # Compatible with torch 2.1
]

for pkg in compatible_versions:
    print(f"   Installing {pkg}...")
    result = subprocess.run(
        [sys.executable, '-m', 'pip', 'install', pkg],
        capture_output=True,
        text=True
    )
    if result.returncode == 0:
        print(f"   ✅ {pkg}")
    else:
        print(f"   ⚠️  {pkg} - check manually")

print()

# Step 3: Install ComfyUI dependencies
print("[3/4] 📦 Installing ComfyUI dependencies...")

comfyui_deps = [
    'einops',
    'safetensors',
    'accelerate', 
    'opencv-python',
    'pillow',
    'scipy',
    'tqdm',
    'kornia',
    'spandrel',
    'soundfile',
    'pyyaml',
    'pycloudflared',
]

for pkg in comfyui_deps:
    subprocess.run(
        [sys.executable, '-m', 'pip', 'install', '-q', pkg],
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL
    )

print("   ✅ All dependencies installed")
print()

# Step 4: Verify installation
print("[4/4] 🔍 Verifying installation...")

try:
    import numpy as np
    import torch
    import transformers
    
    print(f"   ✅ NumPy: {np.__version__}")
    print(f"   ✅ PyTorch: {torch.__version__}")
    print(f"   ✅ Transformers: {transformers.__version__}")
    print(f"   ✅ CUDA available: {torch.cuda.is_available()}")
    
except Exception as e:
    print(f"   ⚠️  Verification issue: {e}")

print()
print("=" * 70)
print("✅ VERSION FIX COMPLETE!")
print("=" * 70)
print()
print("⚠️  IMPORTANT: RESTART THE KERNEL NOW!")
print()
print("Steps:")
print("   1. Go to: Kernel → Restart Kernel")
print("   2. Then run the ComfyUI startup script")
print()
print("=" * 70)
```
