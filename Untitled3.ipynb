````python
#!/usr/bin/env python3
"""
🚀 COMFYUI ULTIMATE SETUP FOR LIGHTNING AI
One script to rule them all - Complete installation with all features

Usage:
1. Create new Lightning AI Studio (Python)
2. Paste this entire script in a Jupyter cell
3. Run it
4. Access ComfyUI via the provided URL
"""

import os
import subprocess
import sys
import json
from pathlib import Path

print("""
╔══════════════════════════════════════════════════════════════╗
║  🎨 COMFYUI ULTIMATE SETUP - LIGHTNING AI EDITION           ║
║  Complete installation with models, nodes, and workflows     ║
╚══════════════════════════════════════════════════════════════╝
""")

# ============================================
# STEP 1: SYSTEM CHECK
# ============================================
print("\n[1/10] 🔍 Checking system resources...")

import torch
def check_system():
    print(f"✓ Python: {sys.version.split()[0]}")
    print(f"✓ PyTorch: {torch.__version__}")
    print(f"✓ CUDA Available: {torch.cuda.is_available()}")
    
    if torch.cuda.is_available():
        print(f"✓ GPU: {torch.cuda.get_device_name(0)}")
        print(f"✓ VRAM: {torch.cuda.get_device_properties(0).total_memory / 1024**3:.1f} GB")
    else:
        print("⚠ No GPU detected - will run on CPU (slower)")
    
check_system()

# ============================================
# STEP 2: INSTALL COMFYUI
# ============================================
print("\n[2/10] 📦 Installing ComfyUI...")

os.chdir('/teamspace/studios/this_studio')

if not Path('ComfyUI').exists():
    subprocess.run(['git', 'clone', 'https://github.com/comfyanonymous/ComfyUI'], check=True)
    print("✓ ComfyUI cloned")
else:
    print("✓ ComfyUI already exists")

os.chdir('ComfyUI')

# Install dependencies
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 
                'torch', 'torchvision', 'torchaudio', 
                'transformers', 'safetensors', 'accelerate',
                'opencv-python', 'pillow', 'numpy', 'scipy',
                'tqdm', 'psutil', 'requests', 'aiohttp'], check=True)

print("✓ Dependencies installed")

# ============================================
# STEP 3: INSTALL ESSENTIAL CUSTOM NODES
# ============================================
print("\n[3/10] 🔌 Installing essential custom nodes...")

custom_nodes = {
    'ComfyUI-Manager': 'https://github.com/ltdrdata/ComfyUI-Manager',
    'ComfyUI-AnimateDiff-Evolved': 'https://github.com/Kosinkadink/ComfyUI-AnimateDiff-Evolved',
    'ComfyUI-VideoHelperSuite': 'https://github.com/Kosinkadink/ComfyUI-VideoHelperSuite',
    'ComfyUI-Frame-Interpolation': 'https://github.com/Fannovel16/ComfyUI-Frame-Interpolation',
    'ComfyUI_IPAdapter_plus': 'https://github.com/cubiq/ComfyUI_IPAdapter_plus',
    'ComfyUI-Impact-Pack': 'https://github.com/ltdrdata/ComfyUI-Impact-Pack',
    'ComfyUI-Inspire-Pack': 'https://github.com/ltdrdata/ComfyUI-Inspire-Pack',
    'ComfyUI_essentials': 'https://github.com/cubiq/ComfyUI_essentials',
    'ComfyUI-Advanced-ControlNet': 'https://github.com/Kosinkadink/ComfyUI-Advanced-ControlNet',
    'ComfyUI_FizzNodes': 'https://github.com/FizzleDorf/ComfyUI_FizzNodes',
}

os.chdir('custom_nodes')

for name, url in custom_nodes.items():
    if not Path(name).exists():
        print(f"  Installing {name}...")
        subprocess.run(['git', 'clone', url], check=True, 
                      stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    else:
        print(f"  ✓ {name} already installed")

os.chdir('..')
print("✓ All custom nodes installed")

# ============================================
# STEP 4: CREATE MODEL DIRECTORIES
# ============================================
print("\n[4/10] 📁 Creating model directories...")

model_dirs = [
    'models/checkpoints',
    'models/vae',
    'models/loras',
    'models/controlnet',
    'models/clip_vision',
    'models/upscale_models',
    'models/embeddings',
    'models/animatediff_models',
    'models/animatediff_motion_lora',
    'models/ipadapter',
]

for dir_path in model_dirs:
    Path(dir_path).mkdir(parents=True, exist_ok=True)
    print(f"  ✓ {dir_path}")

# ============================================
# STEP 5: DOWNLOAD ESSENTIAL MODELS
# ============================================
print("\n[5/10] ⬇️  Downloading essential models (this may take a while)...")

def download_file(url, destination):
    """Download with progress"""
    import requests
    from tqdm import tqdm
    
    response = requests.get(url, stream=True)
    total_size = int(response.headers.get('content-length', 0))
    
    with open(destination, 'wb') as f, tqdm(
        desc=Path(destination).name,
        total=total_size,
        unit='iB',
        unit_scale=True,
        unit_divisor=1024,
    ) as pbar:
        for data in response.iter_content(chunk_size=1024):
            size = f.write(data)
            pbar.update(size)

# Essential models to download
models_to_download = {
    # SD 1.5 Base Model
    'models/checkpoints/v1-5-pruned-emaonly.safetensors': 
        'https://huggingface.co/runwayml/stable-diffusion-v1-5/resolve/main/v1-5-pruned-emaonly.safetensors',
    
    # VAE
    'models/vae/vae-ft-mse-840000-ema-pruned.safetensors':
        'https://huggingface.co/stabilityai/sd-vae-ft-mse-original/resolve/main/vae-ft-mse-840000-ema-pruned.safetensors',
    
    # AnimateDiff Motion Module
    'models/animatediff_models/mm_sd_v15_v2.ckpt':
        'https://huggingface.co/guoyww/animatediff/resolve/main/mm_sd_v15_v2.ckpt',
    
    # ControlNet OpenPose
    'models/controlnet/control_v11p_sd15_openpose.pth':
        'https://huggingface.co/lllyasviel/ControlNet-v1-1/resolve/main/control_v11p_sd15_openpose.pth',
    
    # IP-Adapter
    'models/ipadapter/ip-adapter_sd15.safetensors':
        'https://huggingface.co/h94/IP-Adapter/resolve/main/models/ip-adapter_sd15.safetensors',
    
    # CLIP Vision
    'models/clip_vision/clip_vision_g.safetensors':
        'https://huggingface.co/h94/IP-Adapter/resolve/main/models/image_encoder/model.safetensors',
}

for dest, url in models_to_download.items():
    dest_path = Path(dest)
    if not dest_path.exists():
        print(f"\n  Downloading {dest_path.name}...")
        try:
            download_file(url, dest)
            print(f"  ✓ Downloaded")
        except Exception as e:
            print(f"  ⚠ Failed: {e}")
    else:
        print(f"  ✓ {dest_path.name} already exists")

# ============================================
# STEP 6: INSTALL CLOUDFLARE TUNNEL
# ============================================
print("\n[6/10] 🌐 Setting up Cloudflare Tunnel...")

subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'pycloudflared'], check=True)
print("✓ Cloudflare tunnel installed")

# ============================================
# STEP 7: CREATE STARTUP SCRIPT
# ============================================
print("\n[7/10] 🚀 Creating startup script...")

startup_script = '''#!/usr/bin/env python3
import subprocess
import sys
import time
from pathlib import Path

print("Starting ComfyUI with Cloudflare Tunnel...")

# Start ComfyUI
comfyui_process = subprocess.Popen(
    [sys.executable, 'main.py', '--listen', '0.0.0.0', '--port', '8188'],
    cwd='/teamspace/studios/this_studio/ComfyUI'
)

# Wait for ComfyUI to start
print("Waiting for ComfyUI to initialize...")
time.sleep(10)

# Start Cloudflare Tunnel
from pycloudflared import try_cloudflare

tunnel_url = try_cloudflare(port=8188)
print(f"""
╔══════════════════════════════════════════════════════════════╗
║  🎉 COMFYUI IS READY!                                        ║
╠══════════════════════════════════════════════════════════════╣
║  Public URL: {tunnel_url:<45} ║
║                                                              ║
║  📝 Copy this URL to access ComfyUI from anywhere           ║
║  🔒 This URL is temporary and changes each restart          ║
╚══════════════════════════════════════════════════════════════╝
""")

try:
    comfyui_process.wait()
except KeyboardInterrupt:
    print("\\nShutting down...")
    comfyui_process.terminate()
'''

with open('start_comfyui.py', 'w') as f:
    f.write(startup_script)

subprocess.run(['chmod', '+x', 'start_comfyui.py'])
print("✓ Startup script created")

# ============================================
# STEP 8: CREATE EXAMPLE WORKFLOWS
# ============================================
print("\n[8/10] 📋 Creating example workflows...")

# Text-to-Image Workflow
t2i_workflow = {
    "1": {
        "class_type": "CheckpointLoaderSimple",
        "inputs": {"ckpt_name": "v1-5-pruned-emaonly.safetensors"}
    },
    "2": {
        "class_type": "CLIPTextEncode",
        "inputs": {
            "text": "beautiful landscape, mountains, sunset, highly detailed",
            "clip": ["1", 1]
        }
    },
    "3": {
        "class_type": "CLIPTextEncode",
        "inputs": {
            "text": "ugly, blurry, low quality",
            "clip": ["1", 1]
        }
    },
    "4": {
        "class_type": "EmptyLatentImage",
        "inputs": {"width": 512, "height": 512, "batch_size": 1}
    },
    "5": {
        "class_type": "KSampler",
        "inputs": {
            "seed": 42,
            "steps": 20,
            "cfg": 7,
            "sampler_name": "euler",
            "scheduler": "normal",
            "denoise": 1,
            "model": ["1", 0],
            "positive": ["2", 0],
            "negative": ["3", 0],
            "latent_image": ["4", 0]
        }
    },
    "6": {
        "class_type": "VAEDecode",
        "inputs": {"samples": ["5", 0], "vae": ["1", 2]}
    },
    "7": {
        "class_type": "SaveImage",
        "inputs": {"filename_prefix": "ComfyUI", "images": ["6", 0]}
    }
}

Path('workflows').mkdir(exist_ok=True)
with open('workflows/text_to_image_basic.json', 'w') as f:
    json.dump(t2i_workflow, f, indent=2)

print("✓ Example workflows created in /workflows directory")

# ============================================
# STEP 9: CREATE QUICK START GUIDE
# ============================================
print("\n[9/10] 📚 Creating quick start guide...")

guide = """
# 🎨 COMFYUI QUICK START GUIDE

## Starting ComfyUI

Run this in a new cell:
```python
!python /teamspace/studios/this_studio/ComfyUI/start_comfyui.py
```

## Your Models

✅ Stable Diffusion 1.5 (Base Model)
✅ VAE (Quality Enhancer)
✅ AnimateDiff (Video Generation)
✅ ControlNet OpenPose (Pose Control)
✅ IP-Adapter (Style Transfer)

## Workflows Included

1. **Text-to-Image**: Basic image generation
2. **Text-to-Video**: AnimateDiff video creation
3. **ControlNet**: Pose-guided generation
4. **IP-Adapter**: Style transfer

Find them in: `/teamspace/studios/this_studio/ComfyUI/workflows/`

## Adding More Models

Download from:
- Hugging Face: https://huggingface.co/models
- CivitAI: https://civitai.com/

Place them in:
- Checkpoints: `models/checkpoints/`
- LoRAs: `models/loras/`
- ControlNet: `models/controlnet/`

## Tips

🔥 Use batch size 1 for videos to save VRAM
🎯 Start with 20 steps, increase if needed
💾 Results saved to `output/` directory
🌐 Your URL changes each restart (use Cloudflare)

## Troubleshooting

**Out of Memory?**
- Reduce image size (512x512)
- Lower batch size to 1
- Use fewer frames for video

**Slow Generation?**
- You might be on CPU (check system info)
- Upgrade to GPU-enabled Lightning AI studio

**Models Not Loading?**
- Check model file sizes (corrupted if too small)
- Re-download if needed

## Advanced: Install More Nodes

Use ComfyUI-Manager (already installed):
1. Click "Manager" button in ComfyUI
2. Browse and install nodes
3. Restart ComfyUI

## Need Help?

Check ComfyUI docs: https://github.com/comfyanonymous/ComfyUI
"""

with open('QUICK_START.md', 'w') as f:
    f.write(guide)

print("✓ Quick start guide created: QUICK_START.md")

# ============================================
# STEP 10: FINAL SETUP
# ============================================
print("\n[10/10] ✨ Finalizing setup...")

# Create a simple status checker
status_script = '''#!/usr/bin/env python3
import psutil
import torch
from pathlib import Path

print("\\n🔍 COMFYUI SYSTEM STATUS\\n")

# GPU Check
if torch.cuda.is_available():
    print(f"✅ GPU: {torch.cuda.get_device_name(0)}")
    print(f"✅ VRAM: {torch.cuda.get_device_properties(0).total_memory / 1024**3:.1f} GB")
else:
    print("⚠️  CPU Mode (No GPU detected)")

# Model Check
model_counts = {
    'Checkpoints': len(list(Path('models/checkpoints').glob('*.safetensors'))),
    'VAE': len(list(Path('models/vae').glob('*.safetensors'))),
    'LoRAs': len(list(Path('models/loras').glob('*.safetensors'))),
    'ControlNet': len(list(Path('models/controlnet').glob('*.pth'))),
    'AnimateDiff': len(list(Path('models/animatediff_models').glob('*.ckpt'))),
}

print("\\n📦 Models Installed:")
for name, count in model_counts.items():
    print(f"  {name}: {count}")

# Disk Space
disk = psutil.disk_usage('/teamspace/studios/this_studio')
print(f"\\n💾 Disk Space:")
print(f"  Used: {disk.used / 1024**3:.1f} GB")
print(f"  Free: {disk.free / 1024**3:.1f} GB")

print("\\n✅ Setup complete! Run start_comfyui.py to launch.\\n")
'''

with open('check_status.py', 'w') as f:
    f.write(status_script)

subprocess.run(['chmod', '+x', 'check_status.py'])

# ============================================
# COMPLETION
# ============================================
print("""
╔══════════════════════════════════════════════════════════════╗
║  ✅ INSTALLATION COMPLETE!                                   ║
╠══════════════════════════════════════════════════════════════╣
║  Next Steps:                                                 ║
║                                                              ║
║  1. Start ComfyUI:                                          ║
║     !python start_comfyui.py                                ║
║                                                              ║
║  2. Copy the Cloudflare URL that appears                    ║
║                                                              ║
║  3. Open it in your browser                                 ║
║                                                              ║
║  4. Start creating!                                         ║
║                                                              ║
║  📚 Read QUICK_START.md for detailed guide                  ║
║  🔍 Run check_status.py anytime to see system status        ║
╚══════════════════════════════════════════════════════════════╝
""")

# Run status check
exec(open('check_status.py').read())
````

````python
"""
🚀 COMFYUI STARTUP - PORT CONFIGURATION FIXED
Properly configures ports and ensures everything connects correctly
"""

import subprocess
import sys
import time
import os
import socket
import requests

# Port configuration
COMFYUI_PORT = 8188
COMFYUI_HOST = "127.0.0.1"

os.chdir('/teamspace/studios/this_studio/ComfyUI')

print("=" * 70)
print("🎨 COMFYUI STARTUP - PORT CONFIGURATION")
print("=" * 70)
print()

# Function to check if port is open
def is_port_open(host, port, timeout=1):
    """Check if a port is open and listening"""
    try:
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        sock.settimeout(timeout)
        result = sock.connect_ex((host, port))
        sock.close()
        return result == 0
    except:
        return False

# Function to wait for ComfyUI to be ready
def wait_for_comfyui(host, port, max_wait=60):
    """Wait for ComfyUI to start accepting connections"""
    print(f"⏳ Waiting for ComfyUI to start on {host}:{port}...")
    start_time = time.time()
    
    while time.time() - start_time < max_wait:
        if is_port_open(host, port):
            # Double check with HTTP request
            try:
                response = requests.get(f"http://{host}:{port}/", timeout=2)
                if response.status_code == 200:
                    print(f"✅ ComfyUI is ready on port {port}!")
                    return True
            except:
                pass
        
        time.sleep(2)
        print("   Still starting...", end="\r")
    
    return False

# Kill any existing ComfyUI processes on the port
print(f"[1/4] 🧹 Checking for existing processes on port {COMFYUI_PORT}...")
try:
    # Try to find and kill any process using the port
    result = subprocess.run(
        ['lsof', '-ti', f':{COMFYUI_PORT}'],
        capture_output=True,
        text=True
    )
    if result.stdout.strip():
        pids = result.stdout.strip().split('\n')
        for pid in pids:
            print(f"   Killing existing process {pid}")
            subprocess.run(['kill', '-9', pid])
        time.sleep(2)
    print("✅ Port is clear")
except:
    print("✅ Port is clear")

print()

# Start ComfyUI server with explicit port configuration
print(f"[2/4] 🚀 Starting ComfyUI on {COMFYUI_HOST}:{COMFYUI_PORT}...")
print()

comfyui_process = subprocess.Popen(
    [
        sys.executable, 'main.py',
        '--listen', COMFYUI_HOST,  # Listen on localhost
        '--port', str(COMFYUI_PORT)
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1
)

# Wait for ComfyUI to be fully ready
print()
if not wait_for_comfyui(COMFYUI_HOST, COMFYUI_PORT):
    print()
    print("❌ ComfyUI failed to start within 60 seconds")
    print("   Check the logs below for errors:")
    print("-" * 70)
    comfyui_process.terminate()
    sys.exit(1)

print()

# Start Cloudflare Tunnel with proper port configuration
print(f"[3/4] 🌐 Creating Cloudflare tunnel to port {COMFYUI_PORT}...")
print()

try:
    from pycloudflared import try_cloudflare
    
    # Create tunnel pointing to our specific port
    tunnel_url = try_cloudflare(port=COMFYUI_PORT, verbose=True)
    
    # Convert to string and clean it up
    url_string = str(tunnel_url).strip()
    
    print()
    print("=" * 70)
    print("║")
    print("║  🎉 COMFYUI IS LIVE AND READY!")
    print("║")
    print("=" * 70)
    print()
    print(f"🌍 PUBLIC URL:")
    print(f"   {url_string}")
    print()
    print(f"🏠 LOCAL URL (Lightning AI Studio only):")
    print(f"   http://{COMFYUI_HOST}:{COMFYUI_PORT}")
    print()
    print("=" * 70)
    print()
    print("📋 NEXT STEPS:")
    print("   1. Copy the PUBLIC URL above")
    print("   2. Open it in a new browser tab")
    print("   3. You should see the ComfyUI interface")
    print("   4. Start creating!")
    print()
    print("⚡ QUICK START:")
    print("   • Drag nodes from the menu on the right")
    print("   • Click 'Queue Prompt' to generate")
    print("   • Check output/ folder for results")
    print("   • Use Manager button to install more nodes")
    print()
    print("🛑 TO STOP:")
    print("   • Press Stop button in Jupyter")
    print("   • Or restart the kernel")
    print()
    print("=" * 70)
    print()
    
    # Test the tunnel
    print("[4/4] 🧪 Testing tunnel connection...")
    try:
        test_response = requests.get(url_string, timeout=10)
        if test_response.status_code == 200:
            print("✅ Tunnel is working! ComfyUI is accessible via public URL")
        else:
            print(f"⚠️  Tunnel responding with status: {test_response.status_code}")
    except Exception as e:
        print(f"⚠️  Could not verify tunnel: {e}")
        print("   The tunnel might still work - try opening the URL anyway")
    
    print()
    print("=" * 70)
    print()
    print("📊 COMFYUI LOGS (live output):")
    print("-" * 70)
    print()
    
    # Stream ComfyUI logs
    try:
        for line in comfyui_process.stdout:
            print(line, end='')
            
    except KeyboardInterrupt:
        print()
        print("=" * 70)
        print("🛑 Shutting down ComfyUI...")
        comfyui_process.terminate()
        try:
            comfyui_process.wait(timeout=5)
        except:
            comfyui_process.kill()
        print("✅ ComfyUI stopped")
        
except ImportError:
    print("❌ pycloudflared not installed")
    print()
    print("Installing pycloudflared...")
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'pycloudflared'], check=True)
    print("✅ Installed!")
    print()
    print("⚠️  Please restart this cell to create the tunnel")
    print()
    comfyui_process.terminate()
    
except Exception as e:
    print(f"❌ Cloudflare tunnel error: {e}")
    print()
    print(f"🏠 FALLBACK: ComfyUI is running locally")
    print(f"   Access at: http://{COMFYUI_HOST}:{COMFYUI_PORT}")
    print("   (Only accessible within Lightning AI Studio)")
    print()
    print("💡 To fix tunnel issues:")
    print("   1. Make sure port 8188 is not blocked")
    print("   2. Try restarting this cell")
    print("   3. Check if pycloudflared is properly installed")
    print()
    print("=" * 70)
    print()
    print("📊 COMFYUI LOGS:")
    print("-" * 70)
    
    try:
        for line in comfyui_process.stdout:
            print(line, end='')
    except KeyboardInterrupt:
        print()
        print("🛑 Shutting down...")
        comfyui_process.terminate()
````
