```python
import csv, os, sys, hashlib, shutil, tempfile, urllib.request

MANIFEST = sys.argv[1] if len(sys.argv) > 1 else "models_manifest.csv"
ROOT = os.getcwd().replace("\\", "/")

def ensure_dir(p):
    os.makedirs(p, exist_ok=True)

def human(n):
    for u in ["B","KB","MB","GB","TB"]:
        if n<1024: return f"{n:.1f}{u}"
        n/=1024
    return f"{n:.1f}PB"

def download(url, dst):
    tmp = dst + ".part"
    headers = {}
    pos = 0
    if os.path.exists(tmp):
        pos = os.path.getsize(tmp)
        headers["Range"] = f"bytes={pos}-"
    req = urllib.request.Request(url, headers=headers)
    with urllib.request.urlopen(req) as r, open(tmp, "ab") as f:
        total = r.length + pos if r.length else None
        while True:
            chunk = r.read(1024 * 1024)
            if not chunk: break
            f.write(chunk)
            if total:
                pct = (f.tell() / total) * 100
                print(f"\r→ {os.path.basename(dst)} {pct:5.1f}% ({human(f.tell())}/{human(total)})", end="")
    print()
    os.replace(tmp, dst)

def main():
    with open(MANIFEST, newline="", encoding="utf-8") as f:
        rdr = csv.reader(f)
        for row in rdr:
            if not row or row[0].strip().startswith("#"): continue
            target_dir, url = row[0].strip(), row[1].strip()
            rename = row[2].strip() if len(row) > 2 else ""
            if not url: continue
            target_dir = target_dir.replace("\\", "/")
            ensure_dir(target_dir)
            filename = rename or os.path.basename(url.split("?")[0])
            dest = os.path.join(target_dir, filename)
            if os.path.exists(dest) and os.path.getsize(dest) > 0:
                print(f"✓ Exists: {dest}")
                continue
            print(f"↓ Downloading: {url}\n   → {dest}")
            try:
                download(url, dest)
            except Exception as e:
                print(f"✗ Failed: {url}\n  Reason: {e}")

if __name__ == "__main__":
    main()
```
